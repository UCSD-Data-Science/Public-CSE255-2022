# Principal Component Analysis (PCA) Explained in Matrix Notation


## Introduction

Principal Component Analysis (PCA) is a dimensionality reduction technique that finds a low-dimensional representation of data that captures the most variance. 

In this notebook, we explain PCA using **matrix notation**, step by step.



## Step 1: Center the Data

Given a dataset with $n$ samples and $d$ features, represented by a matrix $X \in \mathbb{R}^{n \times d}$, we first subtract the mean of each column to center the data:

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i, \quad X_{\text{centered}} = X - \mathbf{1}\bar{x}^\top
$$


```python

import numpy as np

# Create a toy dataset
X = np.array([[2.5, 2.4],
              [0.5, 0.7],
              [2.2, 2.9],
              [1.9, 2.2],
              [3.1, 3.0],
              [2.3, 2.7],
              [2.0, 1.6],
              [1.0, 1.1],
              [1.5, 1.6],
              [1.1, 0.9]])

# Center the data
X_mean = np.mean(X, axis=0)
X_centered = X - X_mean

X_centered

```

```python
import matplotlib.
```


## Step 2: Compute the Covariance Matrix

The empirical covariance matrix of the centered data is:

$$
C = \frac{1}{n} X^\top X
$$


```python

# Compute the covariance matrix
n_samples = X_centered.shape[0]
C = (X_centered.T @ X_centered) / n_samples
C
```


## Step 3: Eigen Decomposition

We compute the eigenvalues and eigenvectors of the covariance matrix:

$$
C v = \lambda v
$$

The eigenvectors are the principal components. Sorting by decreasing eigenvalue gives the directions of maximum variance.


```python

# Eigen decomposition
eigvals, eigvecs = np.linalg.eigh(C)

# Sort by eigenvalue in descending order
idx = np.argsort(eigvals)[::-1]
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]

eigvals, eigvecs
```


## Step 4: Project the Data

To reduce dimensionality to $k$ components, project the centered data onto the top $k$ eigenvectors:

$$
Z = X_{\text{centered}} V_k
$$


```python

# Project onto the first principal component
k = 1
V_k = eigvecs[:, :k]
Z = X_centered @ V_k

Z
```


## Step 5: Reconstruction from Principal Components

We can approximately reconstruct the original data from the lower-dimensional representation:

$$
\hat{X} = Z V_k^\top + \bar{x}
$$


```python

# Reconstruct from the projection
X_reconstructed = Z @ V_k.T + X_mean
X_reconstructed
```
