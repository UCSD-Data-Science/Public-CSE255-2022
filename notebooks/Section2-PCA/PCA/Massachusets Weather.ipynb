# Climate of Massachusetts

Massachusetts has a **humid continental climate** with four distinct seasons:

## General Characteristics
- **Winters**: Cold and snowy.
  - Inland areas (e.g., Worcester): 20°F to 25°F (−6°C to −4°C).
  - Coastal areas (e.g., Boston): around 30°F (−1°C).
- **Summers**: Warm to hot and humid.
  - Inland temperatures: around 80°F (27°C).
  - Coastal areas are cooler due to ocean breezes.
- **Spring**: Cool and wet, warming rapidly by May.
- **Fall**: Mild, dry, and famous for colorful foliage.

## Regional Differences
- **Coastal Massachusetts (Boston, Cape Cod, Islands)**:
  - Milder winters and cooler summers.
  - Higher humidity and frequent fog.
- **Central and Western Massachusetts (Worcester Hills, Berkshires)**:
  - Colder winters, heavier snowfall.
  - Cooler nights even during summer.

## Precipitation
- Rainfall is **evenly distributed** throughout the year.
- Annual precipitation: **45–50 inches** (1150–1270 mm).
- Snowfall varies:
  - **30–40 inches** along the coast.
  - **70+ inches** inland and in higher elevations.

## Extreme Weather
- **Nor’easters**: Powerful storms in winter, bringing snow and coastal flooding.
- **Hurricanes/Tropical Storms**: Occasional, mostly between August and October.
- **Summer heat waves** and **periodic droughts** can occur.
